# Northwind Database SQL Analysis

## Objective

The objective of this notebook is to analyze the Northwind database using SQL and Pandas to answer key business questions, explore sales performance, and generate meaningful business insights.

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

## Connect to the Database

```python
conn = sqlite3.connect("northwind.db")
```

## View Available Tables

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

pd.read_sql_query(query, conn)
```

## 1. Top 10 Selling Products

```python
query1 = """
SELECT
    p.ProductName,
    SUM(od.Quantity) AS TotalSold
FROM "Order Details" od
JOIN Products p
ON od.ProductID = p.ProductID
GROUP BY p.ProductName
ORDER BY TotalSold DESC
LIMIT 10;
"""

top_products = pd.read_sql_query(query1, conn)
top_products
```

## 2. Top 10 Customers by Revenue

```python
query2 = """
SELECT
    c.CompanyName,
    ROUND(SUM(od.UnitPrice * od.Quantity * (1 - od.Discount)),2) AS Revenue
FROM Customers c
JOIN Orders o
ON c.CustomerID = o.CustomerID
JOIN "Order Details" od
ON o.OrderID = od.OrderID
GROUP BY c.CompanyName
ORDER BY Revenue DESC
LIMIT 10;
"""

top_customers = pd.read_sql_query(query2, conn)
top_customers
```

## 3. Monthly Sales Trends

```python
query3 = """
SELECT
    strftime('%Y-%m', o.OrderDate) AS Month,
    ROUND(SUM(od.UnitPrice * od.Quantity * (1 - od.Discount)),2) AS Sales
FROM Orders o
JOIN "Order Details" od
ON o.OrderID = od.OrderID
GROUP BY Month
ORDER BY Month;
"""

monthly_sales = pd.read_sql_query(query3, conn)
monthly_sales
```

## 4. Best Performing Product Categories

```python
query4 = """
SELECT
    c.CategoryName,
    ROUND(SUM(od.UnitPrice * od.Quantity * (1 - od.Discount)),2) AS Revenue
FROM Categories c
JOIN Products p
ON c.CategoryID = p.CategoryID
JOIN "Order Details" od
ON p.ProductID = od.ProductID
GROUP BY c.CategoryName
ORDER BY Revenue DESC;
"""

categories = pd.read_sql_query(query4, conn)
categories
```

## 5. Customer Purchase Frequency

```python
query5 = """
SELECT
    c.CompanyName,
    COUNT(o.OrderID) AS OrdersPlaced
FROM Customers c
JOIN Orders o
ON c.CustomerID = o.CustomerID
GROUP BY c.CompanyName
ORDER BY OrdersPlaced DESC;
"""

purchase_frequency = pd.read_sql_query(query5, conn)
purchase_frequency
```
